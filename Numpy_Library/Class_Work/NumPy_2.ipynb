```python
import numpy as np

print("ABC" + "DEF" + "GHI")

a = np.array([[1,2],[3,4],[11,12]])
print(a)

b= np.array([[5,6],[7,8],[11,12]]) 
print(b)

c = np.array([[9,10],[11,12],[11,12]])
print(c)
print("\n")

print("Concatenate :\n",np.concatenate((a,b,c)))
print("Concatenate along axis 0 :\n",np.concatenate((a,b,c), axis=0))
print("Concatenate along axis 1 :\n",np.concatenate((a,b,c), axis=1))
```

```python
# Stacking Functions

print("\n")
print("V-Stack :\n",np.vstack((a,b,c)))
print("\n")
print("H-Stack :\n",np.hstack((a,b,c))) 
print("\n")
print("D-Stack :\n",np.dstack((a,b,c))) 
```

```python
# Split Functions
arr = np.arange(9)
print(arr)
# print("Split : ",np.split(arr,3))

print("Array Split : ",np.array_split(arr,4))
```

```python
# Insert, Append, Delete Functions
arr = np.array([10,20,30])

print("Insert : ",np.insert(arr,1,15))
c = np.insert(arr,1,15)
print(c)

print(arr)
print("Append : ",np.append(arr,[40,25]))

print("Delete : ",np.delete(arr,1))
print(arr)
```

```python
arr = np.array([1,2,3])
print(arr)

# arr[0] = 100

view_arr = arr.view()   
# view_arr[0] = 200
copy_arr = arr.copy()   
print("Original Array : ",arr)
print("View Array : ",view_arr)
print("Copy Array : ",copy_arr)
```
