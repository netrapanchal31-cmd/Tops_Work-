```python
# Array manipulation 
# merge = concatenate()
# differenec between plus and coma
# plus add extra space 
print("ABC" + "DEF" + "GHI")
import numpy as np
a = np.array([[1,2],[3,4],[11,12]])
print("Normal \n", a)
b=np.array([[5,6],[7,8],[11,12]]) 
print("Normal \n",b)
c=np.array([[9,10],[11,12],[11,12]])
print("Normal \n",c)

print("Concatenate \n", np.concatenate((a,b,c)))
print("Concatenate axis=0: \n", np.concatenate((a,b,c), axis=0))
print("Concatenate axis=1: \n", np.concatenate((a,b,c), axis=1))
 
```

```python
#stacking functions
print("Vstack:\n",np.vstack((a,b,c)))
print("Hstack:\n",np.hstack((a,b,c)))
print("Dstack:\n",np.dstack((a,b,c)))
```
