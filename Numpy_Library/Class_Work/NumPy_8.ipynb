```python
import numpy as np 
# np.unique --- unique element remove duplicates 
arr=np.array([1,2,2,3,3,3,4,4,5,5])
unique_values=np.unique(arr)
print("Orginial values:\n",arr)
print("Unique values:\n",unique_values)
```

```python
arr=np.array([40,20,10,30])
sorted_values=np.sort(arr)
print("orginal values:\n",arr)
print("sorted values:\n",sorted_values)
```

```python
#arg sort write indexies according to orginal array 
arr=np.array([30,10,50,20])
sorted_arr=np.sort(arr)
indices=np.argsort(arr)
print("Array:\n",arr)
print("indices:\n",indices)
print("Sorted using indices:\n",arr[indices])
```

```python
#non-zero is the one that write indices of the non-zero            
arr=np.array([0,1,0,2,0,3,0])
print("arrya:\n",arr)
non_zero=np.nonzero(arr)
print("Indices of the non_zero\n",non_zero)
print("Number through indices\n",arr[non_zero])
```

```python
arr=np.array([5,10,11])
print("Any element greater than 10:",np.any(arr<5)) #if any true write true
print("All element greater than 10:",np.all(arr>4)) #if any false write false
```

```python
arr=np.array([1,np.nan,np.inf,-np.inf,5])
print("Array:\n",arr)
print("Isnan:\n",np.isnan(arr)) # only write true when there is nan 
print("isinf:\n",np.isinf(arr)) # positive infite and negative infite write true 
print("isinfinte:\n",np.isfinite(arr)) # true when there is number 
```

```python
arr=np.array([10,20,np.nan,40])
print("Array:\n",arr)
print("mean(normal):\n",np.mean(arr))
print("mean ignoring NaN:\n",np.nanmean(arr))
print("Standard deviation ignoring NaN:\n",np.nanstd(arr))
print("Sum ignoring NaN:\n",np.nansum(arr))
```

```python
#clip - give min and max value - convert that into given range 
arr=np.array([2,5,10,9,11,15,20])
clipped=np.clip(arr,8,12)
print("original Array:",arr)
print("Clipped Array:",clipped)
```
