```python
import numpy as np

np.random.seed(42)

print("Uniform distribution:\n", np.random.rand(2,2))
print("Normal distribution:\n", np.random.randn(2,2))
print("Random integers:\n", np.random.randint(1, 10, size=(2,3)))
print("Random choice:\n", np.random.choice([10,20,30], size=9))
```

```python
# Array properties

arr = np.array([[1,2,3],[4,5,6]])
print("Original array:\n", arr)
print("Shape of array:\n", arr.shape)
print("Size of array:\n", arr.size)
print("Data type of array:\n", arr.dtype)
print("Item size of array:\n", arr.itemsize)
print("Number of dimensions of array:\n", arr.ndim)
print("Number of bytes of array:\n", arr.nbytes)
```

```python
# Reshaping Array Methods

arr = np.arange(1, 10)
print(arr)

reshaped = arr.reshape(3,3)
print("Reshaped Array:\n",reshaped)
```

```python
# Transpose
matrix = np.array([[1,2,3],[4,5,6]])
print("Original: \n",matrix)
print("Transpose Using T: \n",matrix.T)
print("Transpose Using Function: \n",np.transpose(matrix))
```
