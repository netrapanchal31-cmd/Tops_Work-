```python
import numpy as np
# elemet and aggragtion 
# element 
a = np.array([10,20,30])
b= np.array([2,4,5])
print(a+b)  #add
print(a-b) #sub
print(a*b)  # multi
print(a/b) # division
print(a//b) #floor division
print(a**b) # power
print(a%b) #modulus
```

```python
#comparison operators 
a=np.array([5,10,15])
print (a<8) #greater than 8
print(a[a>8]) #greater than 8 
print(a==10) #equal to 10
print(a!=8) #not equal to
```

```python
#logical operators 
a= np.array([5,8,15,20])
cond1 = a>8
cond2 =a<18

print(np.logical_and(cond1,cond2)) # correct any one condition true write true 
print(a[np.logical_and(cond1,cond2)])
print(np.logical_or(cond1,cond2)) #either of condition is true write true or else false
print(a[np.logical_or(cond1,cond2)])
print(np.logical_not(cond1)) # write opposite of the answer 
print(a[np.logical_not(cond1)])
```

```python
#elemet-wise vs aggregate 
arr = np.array([1,2,3,4])
print(arr**2) #elemet wise each element in erray 
print(np.sum(arr)) #aggregate
```

```python
a=np.array([4,9,16])
b=np.array([2,3,4])
c=np.array([2,8,4])

print(np.add(a,b)) #add
z=np.add(a,b)
print(np.add(z,c))
print(np.subtract(a,b)) #sub
print(np.divide(a,b)) #divide
print(np.power(a,2)) #power 
print(np.mod(a,b)) #module
print(np.sqrt(a)) #square root
print(np.exp(a)) #exponential
print(np.log(a)) #natural log
print(np.log10(a)) #log base 10
```

```python
arr=np.array([1.5,2.6,3.2])
print(np.round(arr)) #round
print(np.floor(arr)) #floor
print(np.ceil(arr)) #ceil
```

```python
#statistical operation 
data=np.array([10,24,30,36,50])
print(np.mean(data)) #mean
print(np.median(data)) #median 
print(np.std(data)) #if we want the in normal of variance so use STANDARD DEVIATION
print(np.var(data)) #how far the data is from mean as square  VARIANCE
print(np.min(data)) #minimum
print(np.max(data)) #max
print(np.sum(data)) #sum
print(np.cumsum(data)) #cumulative sum  do sum like 1,2,3,4 = 1,1+2=3,3+2=5,5+4=9
print(np.prod(data)) #product it is kind of multiplication
print(np.percentile(data,90)) #90th percentile 
```
