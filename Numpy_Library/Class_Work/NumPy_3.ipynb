```python
import numpy as np

arr = np.array([1,2,2,3,3,4,5,5])
unique_values = np.unique(arr)

print(unique_values)
print(arr)
```

```python
arr = np.array([30,10,50,20])
sorted_arr = np.sort(arr)
indices = np.argsort(arr)

print(sorted_arr)
print(indices)
print(arr)

print("Indices that sort the array :", indices)
print("Sorting using Indices : ", arr[indices])
```

```python
arr = np.array([0,1,0,2,3,0]) 
print("Array :" , arr)

non_zero_indices = np.nonzero(arr)

print("Indices of  Non Zero Elements :",non_zero_indices)
print("Non Zero Values :", arr[non_zero_indices])
```

```python
arr = np.array([5,10,15])

print("Any Element Greater then 10 : ", np.any(arr>10))
print("Any Element Less then 5 : ", np.any(arr<5))
print("Every Element Greater then 4 : ", np.all(arr>4))
print("Every Element Greater then 5 : ", np.all(arr>5))
```

```python
arr = np.array([1, np.nan, np.inf, -np.inf, 5])

print(arr)

print("Is NaN :", np.isnan(arr))
print("Is Infinite :", np.isinf(arr))
print("Is Finite :", np.isfinite(arr))
```

```python
arr = np.array([10,20,np.nan,40])

print("Array :" ,arr)

print("Mean :" ,np.mean(arr))
print("Mean Ignoring NaN :", np.nanmean(arr))

print("Standard Devistion Ignoring NaN :", np.nanstd(arr))

print("Sum Ignoring NaN :", np.nansum(arr))
```

```python
arr = np.array([2,5,10,9,11,15,20])

clipped = np.clip(arr, 8, 12)

print("Original Array :", arr)
print("Clipped Array :", clipped)
```
