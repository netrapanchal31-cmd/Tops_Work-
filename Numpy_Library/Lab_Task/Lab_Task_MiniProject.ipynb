PART- 1 – Basic Exploration

```python
import numpy as np
np.random.seed(42)
prices = np.random.randint(50, 2001, 500)
quantity = np.random.randint(1, 101, 500)
discount = np.random.randint(0, 41, 500)
ratings = np.round(np.random.uniform(1.0, 5.0, 500), 1)
product_names = ["Laptop", "Phone", "Headphones", "Keyboard", "Mouse","Monitor", "Printer", "Speaker", "Tablet", "Camera"]
products = np.random.choice(product_names, 500)
prices[np.random.choice(500, 15, replace=False)] = 0
ratings[np.random.choice(500, 10, replace=False)] = np.nan
```

```python
print("Prices_shape:\n", prices.shape)
print("Quantity_shape:\n", quantity.shape)
print("Discount_shape:\n", discount.shape)
print("Ratings_shape:\n", ratings.shape)
```

```python
print("Prices_size:\n", prices.size)
print("Quantity_size:\n", quantity.size)
print("Discount_size:\n", discount.size)
print("Ratings_size:\n", ratings.size)
```

```python
print("Prices_Data Type:\n", prices.dtype)
print("Quantity_Data Type:\n", quantity.dtype)
print("Discount_Data Type:\n", discount.dtype)
print("Ratings_Data Type:\n", ratings.dtype)
```

PART -2 -Data Cleaning

```python

ratings = np.array([4.5, np.nan, 3.8, np.nan, 5.0])
prices = np.array([100, 0, 250, 300, 0])
print("Missing Ratings:", np.isnan(ratings).sum())
ratings = np.where(np.isnan(ratings), np.nanmean(ratings), ratings)
print("Updated Ratings:", ratings)
print("Zero Price Count:", np.sum(prices == 0))
prices = np.where(prices == 0, np.mean(prices[prices > 0]), prices)
print("Updated Prices:", prices)
```

PART - 3 - Mathematical Operations

```python
np.random.seed(42)
discount_amount = prices * discount / 100
final_price = prices - discount_amount
revenue = final_price * quantity
print("Discount Amount:\n", discount_amount)
print("\nFinal Selling Price:\n", final_price)
print("\nRevenue:\n", revenue)
```

PART -4-  Statistical Analysis

```python
np.random.seed(42)
revenue = final_price * quantity
print("total_revenue =" ,np.sum(revenue))
print("average_revenue =" ,np.mean(revenue))
print("maximum_revenue = ",np.max(revenue))
print("minimum_revenue =", np.min(revenue))
print("mean_price =", np.mean(prices))
print("median_price =", np.median(prices))
print("std_price =", np.std(prices))
print("std_price =", np.std(prices))
```

PART - 6 - Boolean Indexing

```python
np.random.seed(42)
print("Indices of Discount:",np.where(discount > 30)[0])
print("Indices of Costing:",np.where(prices > 1500)[0])
print("Indices of rating:",np.where(ratings > 4)[0])
print("Indices of quantity:",np.where(quantity > 30)[0])
print("Products costing more than ₹1500:",prices>1500)
print("Products with rating greater than 4:", ratings>4)
print("Products with quantity sold greater than 80:", quantity>80)
print("Products having discount greater than 30%:",discount>30)

```

```python
import numpy as np 
np.random.seed(42)
print("Price more than 1000 and Rating is more than 4:", (prices>1000) & (ratings > 4) )
print("Indices of rating and prices :",np.where(ratings > 4) & (prices > 1000)[0])
print("Discount is more than 25per and Quantity is more tha 50:", (discount>25) & (quantity>50))
print("Indices of rating and prices :",np.where(discount > 25) & (quantity> 50)[0])
print("Price less than ₹300 or Rating less than 2:", (prices<300) | (ratings<2))
```


Part 8 – Sorting

```python
price_indices = np.argsort(prices)

print("Sorted Prices:", prices[price_indices]) # print sorted values
print("Indices:", price_indices)  # print indices 
```

```python
revenue=prices*quantity

revenue_indices = np.argsort(revenue)

print("Sorted revenue:", revenue[revenue_indices])
print("Indices:", revenue_indices)
```

```python
ratings_indices = np.argsort(ratings)

print("Sorted ratings:", ratings[ratings_indices])
print("Indices:", ratings_indices)
```
