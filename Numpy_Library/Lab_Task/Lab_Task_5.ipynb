```python
import numpy as np
arr=np.arange(15)
print("Split:",np.split(arr,5)) 
```

```python
arr=np.arange(12)
print("Array Split: \n",np.array_split(arr,4))
```

```python
arr=np.array([[1,2],[3,4]])
print("Insert:\n",np.insert(arr,1,[5,6]))
```

```python
arr=np.array([100,200,300,400,500])
print("Delete:\n",np.delete(arr,3))
print("Delete:\n",np.delete(arr,[0,1]))
```

```python
arr=np.array([[1,2],[3,4],[5,6]])
print("Delete 2 ROW:\n",np.delete(arr,1,axis=0))
```

```python
arr=np.array([1,2,3,4])
view_arr=arr.view()
copy_arr=arr.copy()
view_arr[0]=999
print("Orginial:\n",arr)
print("View:\n",view_arr)
print("Copy:\n",copy_arr)
```

```python
arr=np.arange(1,7)
arr = arr.reshape(2, 3)
arr[0,1]=100 
print(arr)
```

```python
arr=np.array([[1,2,3],[4,5,6]])
faltten=arr.flatten()
ravel=arr.ravel()
faltten[0]=100
ravel[0]=200
print("Flatten:\n",faltten)
print("Ravel:\n",ravel)
```
