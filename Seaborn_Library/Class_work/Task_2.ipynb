# Count Plot
- It is a kind of barplot that plots the value counts of cataegories of a categorical column.

```python
df.sex.value_counts()
```

```python
plt.figure(figsize=(10,8), dpi=100)
a=sns.countplot(x='sex', hue='pclass', data=df, palette='bright')

for i in a.containers:
    a.bar_label(i)

plt.show()
```

```python
df.groupby('sex').pclass.value_counts()
```

# Barplot
- sns.barplot(x=categorical, y=numerical, hue=category, data=df)

```python
a=sns.barplot(x='sex', y='fare', data=df, hue='pclass', errorbar=('ci', False), estimator='median', palette='bright')

for i in a.containers:
    a.bar_label(i, fmt='%.2f')
plt.show()
```
