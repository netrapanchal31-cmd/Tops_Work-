# 3. Why use sampling for Swiggy?
Sampling saves **time, cost, and resources** because collecting delivery-time data from every order would be very large and difficult to manage. A properly selected sample can give us a good idea about overall delivery times without analyzing every order.

# 4. Given this scenario: A survey about music preferences is only sent to users who recently streamed Bollywood songs on Spotify. Identify and explain the type of bias this introduces in the sample.
Convenience Bias: The survey is only sent to users who recently streamed Bollywood songs, making the sample convenient but not representative of all Spotify users.

# 5. Choose any two types of sampling (for example: random sampling, stratified sampling, systematic sampling, cluster sampling). For each, describe how you would use it to select a sample of users from BookMyShow to study movie booking habits.
Random Sampling: Randomly select BookMyShow users from the complete list of users so that every user has an equal chance of being selected. Then study their movie booking habits.

Stratified Sampling: Divide BookMyShow users into groups based on categories such as age or location. Then randomly select users from each group so that all groups are properly represented in the sample.
