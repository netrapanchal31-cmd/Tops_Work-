```python
import numpy as np

arr = np.array([4500, 7000, 5000, 8000, 4000, 9000, 6000])
ans = np.var(arr)
print("Variance:", ans)
```

```python
def calculate_standard_deviation(scores):
    return round(np.std(scores), 2)
```

```python
A = np.array([200, 200, 200, 200, 200])
B = np.array([100, 300, 150, 400, 50])
print(np.std(A))
print(np.std(B))

# Friend A is more consistent as he spend 200 daily so the std is 0. However, the Friend B std is 130 which shows the difference in variation
```

```python
salary = np.array([28000, 29000, 27000, 60000, 26500, 27500])
v = np.var(salary)
sd = np.std(salary)
print("Variance:", v)
print("Standard Deviation:", round(sd, 2))
```
