```python
import pandas as pd 

plays = pd.Series([120, 135, 150, 200, 120, 90, 200])
mean_plays = plays.mean()
print("Mean Spotify plays:", mean_plays)
```

```python
delivery_times = pd.Series([30, 25, 40, 35, 30, 45, 30])
median_time = delivery_times.median()
print("Median delivery time:", median_time)
```

```python
def most_common_rating(ratings):
    ratings = pd.Series(ratings)
    return ratings.mode()[0]
ratings = [5, 4, 4, 3, 5, 4, 2, 4]
print("Most common rating:", most_common_rating(ratings))
```

```python
channel_1 = pd.Series([100, 120, 110, 105, 115])
channel_2 = pd.Series([200, 210, 205, 215, 220])
channel_3 = pd.Series([100, 120, 110, 105, 5000])
channels = {
    "Channel 1": channel_1,
    "Channel 2": channel_2,
    "Channel 3": channel_3
}

for name, data in channels.items():
    print("\n", name)
    print("Mean:", data.mean())
    print("Median:", data.median())
    print("Mode:", data.mode().tolist())
```
