```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
arr=np.array([12, 15, 17, 14, 13, 16, 200, 18, 14, 15])
print(arr)
```

```python
sns.kdeplot(arr)
plt.show()
```

The data is right-skewed (positively skewed).
Most of the days have orders between 12 and 18, but on Day 7 the orders suddenly increase to 200. This unusual high value creates a long tail on the right side of the graph, so the data is right-skewed.

```python
Instagram_influencer =np.array([5000, 7000, 8000, 8000, 9000, 10000, 12000, 15000, 95000])
mean=np.mean(Instagram_influencer)
median=np.median(Instagram_influencer)
print("Mean:", mean)
print("Median: ", median)
if mean > median:
    print("Mean is greater than median")
    print("The data is right skewed (Positive skewed)")
elif mean < median:
    print("Mean is lower than median")
    print("The data is left skewed (Negative skewed)")
else:
    print("Mean and median are equal")
    print("The data is balanced")
```

```python
views = [100, 150, 200, 250, 300, 350, 500, 700, 1000, 10000]
mean = np.mean(views)
print("Mean:", mean)
sns.kdeplot(views)
plt.show()
```

Videos had low view in past however one video reached towards 10,000 which makes the bigger difference that why mean is greater than median and the image showcase the positive skewed (right skewed) 

```python
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

sns.kdeplot(ratings)

plt.title("KDE Plot of Flipkart Product Ratings")
plt.show()
```

The distribution is left-skewed (negative skewness) because most ratings are 5, while fewer ratings are 3, creating a longer tail on the left side.
