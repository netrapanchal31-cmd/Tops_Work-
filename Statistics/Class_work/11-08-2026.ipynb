# 1. What is Data?
Data means raw facts, figures, observations, or information collected for a particular purpose.
Data can be numerical or categorical.

# 2. What is Statistics?
Statistics is the science of collecting, organizing, processing, analyzing, interpreting, and presenting data to make informed decisions.
Statistics converts raw data into meaningful information that can be used for decision-making.
    Step 1: Data Collection
            Collect raw data from different sources.
    Step 2: Organization
            After collecting raw data, we organize it into a structured format.
            like : Table, Excel sheets, Dataframes, Databases
    Step 3: Data Processing
            We clean and prepare the data for analysis.
            like : Handling missing values, Removing duplicates, Correcting errors.
    Step 4: Data Analysis
            We analyze the processed data to find patterns, relationships, and useful information.
            like : mean, median, mode, aggregations, correlation, trends
    Step 5: Data Interpretation
            We explain what the analyzed results actually mean.
    Step 6: Data Presentation
            We present the results in an easy-to-understand format.
            like: charts, graphs, tables, dashboards, kpis, powepoint presentations.
    Step 7: Decision Making
            The information obtained from the analysis is used to make informed decisions.
    Step 8: Monitoring & Governance
            After making a decision, we continue to monitor the data and maintain its quality.

# 3. Types of Statistics
There are three important areas you should remember:
    A. Descriptive Statistics
          Descriptive statistics are used to summarize and describe data. 
          Mean - Average 
          Median - Middle value
          Mode - Most frequently occuring value 
    B. Measures of Dispersion
          Dispersion tells us how spread out or scattered the data is.  
          Range
          Variance
          Standard Deviation
          IQR (Interquartile Range) 
    C. Inferential Statistics
          Inferential statistics are used to make conclusions or predictions about a population using sample data.
          Hypothesis Testing
          t-test
          z-test
          ANOVA
          Chi-square test

# 5. Numerical Data
Numerical data represents numbers or measurable quantities.
It has Two types: 
      A. Discrete Data
            Discrete data consists of countable whole numbers (integers).
            You generally get discrete data by counting. 
      B. Continuous Data
            Continuous data can take decimal values and is generally obtained through measurement.

# 6. Categorical Data
Categorical data represents groups, labels, or categories rather than numerical measurements.
It has two main types:
       A. Nominal Data
             Nominal data consists of categories that do not have any natural order or ranking.
       B. Ordinal Data
             Ordinal data consists of categories that have a meaningful order or ranking.
             Like: Low, Medium, High
                   Poor, Average, Good

# 8.Population vs Sample AND Random Sampling
Population
    Population = The entire group or universe that we are interested in studying.
    Example:
    Suppose we want to study the salaries of all employees in a company.
    If the company has 10,000 employees:
    Population: All 10,000 employees.

Sample
    Sample = A smaller subset selected from the population for analysis.
    Instead of asking all 10,000 employees, we may select 500 employees.
    It is a part of the population 

Random Sampling
    A sample should ideally be selected in a way that represents the population.
    Example:
     If a college has: 60% Female and 40% Male 
     and we randomly select a sample of 100 students, the sample should ideally have a reasonably similar representation. 
     A good sample should be representative of the population.
     Random sampling helps reduce selection bias and gives each population member a fair chance of being selected.

```python
import numpy as np
np.random.seed(123)
population = np.random.randint(0,100,size=1000)
print(population)
```

```python
np.random.seed(20)
sample = np.random.choice(population,size=100)
print(sample)
```
