
# Session 9 -- Normal Distrubitution 
1. Normal Distribution
Normal distribution is an ideal distribution of data.
It is bell-shaped and symmetrical.
Data is mostly concentrated around the center.
In a perfectly normal distribution:
# Mean = Median = Mode

2. Central Tendency
Central tendency tells us the center or typical value of the data.The three measures of central tendency are:
Mean
Median
Mode

3. Mean
Mean = Average of all values
Formula: Mean = Σx / n
Where:
Σx = sum of all values
n = number of values
Example: Data: 10, 20, 30, 40, 50
Mean = (10 + 20 + 30 + 40 + 50) / 5
Mean = 30

4. Median
Median = Middle value after arranging the data in ascending order.
Odd number of values: take the middle value.
Even number of values: take the average of the two middle values.
Example: 1, 5, 6, 8, 9
Median = 6

5. Mode
Mode = Most frequently occurring value.
Example: 1, 2, 2, 3, 4
Mode = 2
Two modes → Bimodal
More than two modes → Multimodal
No repeated value → No mode

6. Central Limit Theorem (CLT)
CLT does NOT mean that mean, median and mode are always equal.
CLT says: When we take sufficiently large random samples from a population, the distribution of sample means tends to become approximately normal.
Remember: Population → Samples → Sample Means → Approximately Normal Distribution

7. Dispersion
Dispersion = how much the data is spread out.
Main measures:
Range
Variance
Standard Deviation (SD)
Higher SD/Variance → More spread
Lower SD/Variance → Less spread
Easy Difference:
Central Tendency → Where is the center?
Dispersion → How spread out is the data?
