# What is Pandas?
Pandas is a Python library used for data manipulation, data processing, and data analysis.
The name Pandas comes from "Panel Data", which refers to structured data arranged in rows and columns.
Pandas is mainly used to work with structured/tabular data.

# What kind of data can Pandas handle?
Pandas can work with data from various sources, such as:

CSV files
Excel files
JSON files
SQL databases

# How is Pandas used?
1. Data Processing

Cleaning data
Handling missing values
Removing duplicates
Handling incorrect data
Encoding categorical data
Transforming data

2. Data Analytics

Querying and filtering data
Sorting data
Grouping and aggregating data
Finding meaningful patterns and insights

# Why is Pandas popular?
It can efficiently work with large datasets, including datasets with 100,000+ rows.
It provides easy-to-use functions for manipulating and analyzing data.
It integrates well with other Python libraries such as NumPy, Matplotlib, and Seaborn.

Pandas = A Python tool used to load, clean, manipulate, analyze, and understand structured data.

```python
df = pd.DataFrame({
    "a": [100,200,300],
    "b": [400,500,600],
    "c": [700,800,900],
    "d": [1000,1100,1200]

})
df
```

Pandas has 2 types of Indexes
1 Implicit Index ==> Python Internal Indexes
  - df [Start : Stop : Step]
  - [stop - excluded, start default - 0,    Step default - 1 ]
2 Explicit Index ==> Labelled Indexes
  - - df [Start : Stop : Step]
  - [stop - included, start default -start value of lablled index, Step default - 1 ]

  - df.loc() [Location means Index]
