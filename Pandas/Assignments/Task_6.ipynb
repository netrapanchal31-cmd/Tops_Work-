```python
import pandas as pd
import numpy as np 

file = r"C:\Users\DELL\Downloads\ipl_match_details.csv\ipl_match_details.csv"
df = pd.read_csv(file)
df= df.replace("----", np.nan)
print("First 5 rows", df.head())
print("\nMissing values using isnull():")
print(df.isnull().sum())

print("\nNon-missing values using notnull():")
print(df.notnull().sum())
```

```python
print("Shape BEFORE dropna:", df.shape)
df_clean = df.dropna(axis=0, how='any')

print("Shape AFTER dropna:", df_clean.shape)
```

```python
df['batsman_runs'] = pd.to_numeric(df['batsman_runs'], errors='coerce')
mean_runs = df['batsman_runs'].mean()
print("Mean batsman runs:", mean_runs)
df['batsman_runs'] = df['batsman_runs'].fillna(mean_runs)

print("\nUpdated batsman_runs column:")
print(df['batsman_runs'])
```
