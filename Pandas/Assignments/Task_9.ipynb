```python
import pandas as pd 
timestamps = [
    "2024-06-01 14:30",
    "2024-06-02 09:15",
    "2024-06-03 20:45"
]

dates = pd.to_datetime(timestamps)

print(dates)
```

```python
df = pd.DataFrame({
    "order_date": [
        "2024-06-01 14:30:00",
        "2024-06-02 09:15:00",
        "2024-06-03 20:45:00"
    ]
})

df["order_date"] = pd.to_datetime(df["order_date"])

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["weekday"] = df["order_date"].dt.day_name()

print(df)
```

```python
df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "customer": ["A", "B", "C", "D", "E", "F", "G"],
    "order_date": [
        "2024-06-01 10:30:00",
        "2024-06-02 12:15:00",
        "2024-06-05 09:45:00",
        "2024-06-08 18:20:00",
        "2024-06-10 14:10:00",
        "2024-06-11 16:00:00",
        "2024-06-15 11:30:00"
    ]
})

df["order_date"] = pd.to_datetime(df["order_date"])

df.set_index("order_date", inplace=True)

weekly_orders = df.resample("W").size()

print("Weekly Orders:")
print(weekly_orders)
```

```python
df = pd.DataFrame({
    "post_id": [101, 102, 103, 104, 105, 106],
    "posted_at": [
        "2024-06-01 08:00:00",
        "2024-06-02 12:30:00",
        "2024-06-03 18:45:00",
        "2024-06-04 09:15:00",
        "2024-06-05 22:00:00",
        "2024-06-06 14:10:00"
    ]
})

df["posted_at"] = pd.to_datetime(df["posted_at"], utc=True)

df["posted_at"] = df["posted_at"].dt.tz_convert("Asia/Kolkata")

print(df.head())
```

```python
df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "order_date": [
        "2024-06-01",  # Saturday
        "2024-06-02",  # Sunday
        "2024-06-03",  # Monday
        "2024-06-04",  # Tuesday
        "2024-06-08"   # Saturday
    ]
})

# Convert to datetime
df["order_date"] = pd.to_datetime(df["order_date"])

# Create is_weekend column
df["is_weekend"] = df["order_date"].dt.dayofweek >= 5

print(df)
```
