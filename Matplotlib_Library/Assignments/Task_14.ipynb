```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\DELL\Downloads\ipl_match_details.csv\ipl_match_details.csv")
print(df.head())
print(df["total_runs"].head())

mean = df["total_runs"].mean()
median = df["total_runs"].median()
minimum = df["total_runs"].min()
maximum = df["total_runs"].max()
std = df["total_runs"].std()

print("Mean:", mean)
print("Median:", median)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Standard Deviation:", std)
```

```python
df = pd.read_csv(r"C:\Users\DELL\Downloads\flipkart_com-ecommerce_sample.csv (1)\flipkart_com-ecommerce_sample.csv")
rating_counts = df['retail_price'].value_counts().sort_index()
plt.figure(figsize=(8, 5))

sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values
)

sns.barplot(x=rating_counts.index, y=rating_counts.values)

plt.title('Number of Flipkart Reviews by Rating')
plt.xlabel('Rating (Stars)')
plt.ylabel('Number of Reviews')
plt.show()
```

```python
data = {
    'restaurant': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'average_cost_for_two': [300, 500, 700, 1000, 250, 450, 800, 1200],
    'user_rating': [3.2, 3.5, 4.0, 4.3, 3.0, 3.8, 4.1, 4.5]
}

df = pd.DataFrame(data)

sns.scatterplot(
    data=df,
    x='average_cost_for_two',
    y='user_rating'
)

plt.title('Restaurant Cost vs User Rating')
plt.xlabel('Average Cost for Two')
plt.ylabel('User Rating')

plt.show()
```

```python
data = {
    'movie': [
        'Jawan', 'Pathaan', 'Dangal',
        'KGF 2', 'RRR', '3 Idiots',
        'Animal', 'War', 'Drishyam 2'
    ],
    'genre': [
        'Action', 'Action', 'Drama',
        'Action', 'Drama', 'Comedy',
        'Drama', 'Action', 'Thriller'
    ],
    'box_office_collection': [
        1100, 1050, 2000,
        1200, 1300, 202,
        900, 475, 345
    ]
}

df = pd.DataFrame(data)

print(df)
avg_collection = (
    df.groupby('genre')['box_office_collection']
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

print(avg_collection)
```
