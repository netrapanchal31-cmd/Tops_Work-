```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
numeric_data = tips.select_dtypes(include="number")
sns.pairplot(numeric_data)

plt.show()
```

```python
flights = sns.load_dataset("flights")
flights_pivot = flights.pivot_table(
    index="month",
    columns="year",
    values="passengers"
)
plt.figure(figsize=(12, 6))

sns.heatmap(
    flights_pivot,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd"
)


plt.title("Monthly Passenger Count by Year")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```

```python
fmri = sns.load_dataset("fmri")
sns.relplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    kind="line"
)
plt.title("FMRI Signal Changes Over Time")
plt.xlabel("Timepoint")
plt.ylabel("Signal")

plt.show()
```

```python
titanic = sns.load_dataset("titanic")
sns.catplot(
    data=titanic,
    x="sex",
    y="survived",
    kind="bar",
    errorbar="ci"
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

```python
penguins = sns.load_dataset("penguins")
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="flipper_length_mm",
    kind="reg"
)

plt.show()
```
