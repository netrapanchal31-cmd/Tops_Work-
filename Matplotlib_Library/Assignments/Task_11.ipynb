```python
import matplotlib.pyplot as plt
#company data 

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [120, 135, 150, 145, 170, 190]

departments = ["IT", "HR", "Sales", "Finance"]
employees = [25, 10, 30, 15]

ad_spend = [10, 15, 20, 25, 30, 35]
sales = [80, 95, 110, 125, 145, 160]

products = ["Software", "Hardware", "Cloud", "Support"]
product_revenue = [40, 25, 20, 15]

fig, ax = plt.subplots(2, 2, figsize=(12, 8))
ax[0,0].plot(
    months,
    revenue,
    marker="o",
    color="Blue"
    )
ax[0, 0].set_title("Monthly Revenue")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Revenue (₹ Lakhs)")

ax[0,1].bar(
    departments,
    employees,
    color="orange"
)
ax[0,1].set_title("Employees by Department")
ax[0, 1].set_xlabel("Department")
ax[0, 1].set_ylabel("Number of Employees")

ax[1, 0].scatter(
    ad_spend,
    sales,
    color="green"
)
ax[1, 0].set_title("Advertising Spend vs Sales")
ax[1, 0].set_xlabel("Advertising Spend (₹ Lakhs)")
ax[1, 0].set_ylabel("Sales (₹ Lakhs)")

ax[1, 1].pie(
    product_revenue,
    labels=products,
    autopct="%1.1f%%"
)
ax[1, 1].set_title("Revenue by Product")
plt.suptitle("TechNova Solutions - Business Dashboard", fontsize=16)

plt.tight_layout()

plt.show()
```

```python
influencers = ["Raj Shamani", "Prajakta Koli", "Rohit Sharma", "Virat Kholi", "PM.Modi"]

followers = [250, 180, 220, 200, 270]       # in millions
daily_posts = [3, 4, 2, 2, 1]               # average posts per day
fig, ax1 = plt.subplots(figsize=(10, 6))
ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers (Millions)"
)
ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    label="Average Daily Posts"
)
ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")
plt.title("Instagram Followers vs Average Daily Posts")
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
series = [
    "The Summer I Turned Pretty",
    "Never Have I Ever",
    "XO, Kitty",
    "Ginny & Georgia",
    "My Life with the Walter Boys"
]
viewers = [85, 95, 75, 120, 80]       # viewers in lakhs
imdb_rating = [7.8, 7.8, 6.7, 7.4, 7.4]
plt.figure(figsize=(11, 6))
plt.scatter(
    viewers,
    imdb_rating,
    color="purple",
    marker="o",
    s=100
)
for i in range(len(series)):
    plt.annotate(
        series[i],
        (viewers[i], imdb_rating[i]),
        xytext=(5, 8),
        textcoords="offset points"
    )

plt.ylabel("IMDb Rating")
plt.title("Web Series: Viewers vs IMDb Ratings")
plt.tight_layout()

plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [120, 135, 128, 145, 150, 160,
         155, 170, 180, 250, 210, 190]
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(months, sales, marker='o', label="Monthly Sales")
max_sales = max(sales)
max_index = sales.index(max_sales)
ax.annotate(
    "Big Billion Days",
    xy=(months[max_index], max_sales),
    xytext=(months[max_index], max_sales + 30),
    arrowprops=dict(arrowstyle="->"),
    fontsize=12,
    fontweight="bold"
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()

plt.show()
```
