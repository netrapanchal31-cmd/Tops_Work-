```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)
delivery_times = np.random.randint(20, 61, 50)
sns.histplot(delivery_times, bins=8, kde=True)

plt.title("Distribution of Zomato Delivery Times")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")

plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Total Bill Distribution by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")

plt.show()
```

```python
np.random.seed(42)
teams = [
    "CSK", "MI", "RCB", "KKR",
    "RR", "SRH", "DC", "GT"
]

data = []

for team in teams:
    scores = np.random.normal(loc=170, scale=25, size=50)
    scores = np.clip(scores, 80, 250)

    for score in scores:
        data.append([team, score])
        
df = pd.DataFrame(data, columns=["Team", "Runs"])
sns.set_theme(style="darkgrid")
plt.figure(figsize=(12, 6))

sns.violinplot(
    x="Team",
    y="Runs",
    data=df
)

plt.title("IPL Run Distribution by Team")
plt.xlabel("IPL Team")
plt.ylabel("Runs Scored")

plt.show()

    
```

```python
genres = [
    "Pop", "Pop", "Rock", "Hip-Hop", "Pop",
    "R&B", "Rock", "Pop", "Hip-Hop", "EDM",
    "Pop", "Rock", "R&B", "Hip-Hop", "Pop",
    "EDM", "Rock", "Pop", "Hip-Hop", "R&B",
    "Pop", "EDM", "Rock", "Pop", "Hip-Hop",
    "R&B", "Pop", "Rock", "EDM", "Pop",
    "Hip-Hop", "R&B", "Pop", "Rock", "EDM",
    "Pop", "Hip-Hop", "R&B", "Rock", "Pop"
]

df = pd.DataFrame({"Genre": genres})
plt.figure(figsize=(8, 5))

sns.countplot(
    x="Genre",
    data=df
)

plt.title("Number of Spotify Tracks by Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")

plt.show()
```

```python
steps = [6500, 8200, 7100, 9500, 8800, 10500, 7600]
sns.set_theme(style="whitegrid")
sns.kdeplot(
    steps,
    fill=True,
    color="purple",
    linewidth=2
)
plt.title("Distribution of Daily Step Counts")
plt.xlabel("Number of Steps")
plt.ylabel("Density")

plt.show()
```
