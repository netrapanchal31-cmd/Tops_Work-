```python
import matplotlib.pyplot as plt
days = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

steps = [6500, 7200, 5800, 8100, 7500, 9200, 6800]

plt.figure(figsize=(10, 5))

plt.plot(days, steps, marker='o')

plt.title("Daily Steps - Last 7 Days")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

plt.grid(True)
plt.tight_layout()
plt.savefig("steps_lineplot.png")

plt.show()
```

```python
restaurants = [
    "Agashiye",
    "Gordhan Thal",
    "The Green House",
    "650 - The Global Kitchen",
    "Swati Snacks",
    "Vishalla",
    "Cryo Lab",
    "Millets of Mewar",
    "Tomato's",
    "Honest"
]
ratings = [4.5, 4.4, 4.3, 4.2, 4.5, 4.3, 4.1, 4.2, 4.0, 4.1]
prices = [1200, 700, 900, 1000, 800, 600, 500, 700, 650, 400]
plt.figure(figsize=(10,6))
plt.scatter(ratings,prices)
plt.title("Zomato Ratings vs Average Meal Price - Ahmedabad")
plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price (₹)")
plt.grid(True)
plt.show()
```

```python
apps = ["Swiggy", "Zomato", "Domino's"]
orders = [12, 8, 5]
colors = ["orange", "red", "blue"]
plt.figure(figsize=(8,5))
plt.bar(apps, orders, color=colors)
plt.title("Food Orders in the Last Month")
plt.xlabel("Food Delivery Platform")
plt.ylabel("Number of Orders")
plt.bar([], [], color="orange", label="Swiggy")
plt.bar([], [], color="red", label="Zomato")
plt.bar([], [], color="blue", label="Domino's")
plt.legend()
plt.show()
```

```python
durations = [
    25, 40, 35, 60, 45,
    30, 75, 50, 20, 55,
    65, 40, 35, 80, 45,
    30, 70, 50, 25, 60
]
plt.hist(durations, bins=5, color='pink', edgecolor='black')
plt.title("Spotify Listening Session Durations")
plt.xlabel("Listening Duration (minutes)")
plt.ylabel("Number of Sessions")
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
screen_time = [45, 60, 50, 70, 55, 90, 80]   # minutes
liked_posts = [20, 35, 25, 40, 30, 50, 45]    # posts
fig, ax = plt.subplots(2, 1, figsize=(8, 8))
ax[0].plot(days, screen_time, marker='o', color='blue')

ax[0].set_title("Daily Instagram Screen Time")
ax[0].set_xlabel("Day")
ax[0].set_ylabel("Screen Time (minutes)")
ax[1].bar(days, liked_posts, color='orange')

ax[1].set_title("Posts Liked Each Day")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Number of Posts")
plt.tight_layout()
plt.savefig("social_media_usage.png")
plt.show()
```
