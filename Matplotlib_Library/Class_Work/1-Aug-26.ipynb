```python
import matplotlib.pyplot as plt
import numpy as np 
x = np.arange(1, 11)
y = x ** 3
print("X array: ",x)
print("Y array: ",y)
```

```python
plt.plot(x, y)

plt.show()
```

```python
plt.plot(x, y)

plt.title("X vs Y Relationship" , 
          fontdict={ 'color': 'green', 'size': 25, 'weight': 'bold', 
                    'font':'cursive' }, loc='center')

plt.xlabel('Array-X', fontdict={'color':'red', 'size':15, 
                                'weight':'bold', 
                                'font':'cursive'}, loc='center')
plt.ylabel('Array-Y', fontdict={'color':'red', 'size':15, 
                                'weight':'bold', 'font':'cursive'}, 
                                loc='center')
plt.show()
```

```python

plt.plot(x,y,
        color='r',
        marker='o',
        markersize=10,
        markerfacecolor='yellow',
        markeredgecolor='green',
        markeredgewidth=2,
        ls='--',
        lw=1.8)

plt.show()
```

```python
plt.plot(x,y,
        color='g',
        marker='o',
        markersize=10,
        markerfacecolor='yellow',
        markeredgecolor='red',
        markeredgewidth=2,
        ls='--',
        lw=1.8,
        label='Data trend')

# loc= 0 to 10 or 'upper left', 'upper center'

plt.title('Plot-A', fontdict={'color':'r', 'size':22}, loc='center')

plt.legend()#loc='upper left', labelcolor='m', fontsize=15)

plt.show()
```
