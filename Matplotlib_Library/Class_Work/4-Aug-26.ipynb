```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 11)
y = x ** 3
print("X array: ", x)
print("Y array: ", y)
```

```python
plt.plot(x, y)
plt.show()
```

```python
# Creating a Figure

plt.figure()
```

```python
# Creating a axes

y*3
```

```python
f = plt.figure(figsize=(10,8))

a = f.add_axes([0.1, 0.1, 0.5, 0.5]) # left, bottom, width, height (range 0 to 1)
b = f.add_axes([0.7, 0.1, 0.5, 0.5]) # left, bottom, width, height (range 0 to 1)
c = f.add_axes([0.7, 0.7, 0.5, 0.5]) # left, bottom, width, height (range 0 to 1)
d = f.add_axes([0.1, 0.7, 0.5, 0.5]) # left, bottom, width, height (range 0 to 1)


a.plot(x, y)
b.plot(y,x)
c.plot(x**2, y/2)
d.plot(x/2, y*3)

plt.show()
```

```python
# Axes

f,a = plt.subplots(1,2,figsize=(15,6))
a[0].plot(x,y)
a[1].plot(y,x)

plt.show()
```

```python
# 4 Plots

_,a = plt.subplots(2,2)

a[0,0].plot(x,y)
a[0,1].plot(x**2,y**2)
a[1,0].plot(x/2,y**2)
a[1,1].plot(y,x)
```

```python
# 2 axes

f,a = plt.subplots(1,2,figsize=(15,6), dpi=600)

a[0].plot(x,y, color='g', marker='o', markerfacecolor='yellow', markersize=15)
a[0].set_title('Plot-A', fontdict={'color':'red', 'size':25, 'weight':'bold'})
a[0].set_xlabel('Axis-A',fontdict={'color':'red', 'size':25, 'weight':'bold'})
a[0].set_ylabel('Axis-B',fontdict={'color':'red', 'size':25, 'weight':'bold'})

a[1].plot(y,x, color='red', marker='o', markerfacecolor='yellow', markersize=15)
a[1].set_title('Plot-B',fontdict={'color':'red', 'size':25, 'weight':'bold'})
a[1].set_xlabel('Axis-A',fontdict={'color':'red', 'size':25, 'weight':'bold'})
a[1].set_ylabel('Axis-B',fontdict={'color':'red', 'size':25, 'weight':'bold'})


plt.tight_layout()

plt.show()
```

# PIE CHART

```python
expenses=[3900, 5500, 3700, 4200]
names=['Krish', 'Netra', 'Lata', 'Dhruvil']
```

```python
plt.figure(figsize=(8,8), dpi=300)

plt.pie(expenses, labels=names, autopct='%1.2f%%', 
        colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'],
        explode=(0, 0.1, 0, 0),
        shadow=False,
        startangle=90, 
        pctdistance=0.8,
        labeldistance=0.5)

plt.legend()
plt.show()
```
